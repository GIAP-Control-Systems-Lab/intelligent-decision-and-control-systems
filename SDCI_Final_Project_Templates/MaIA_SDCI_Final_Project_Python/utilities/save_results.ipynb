Saves the metrics table and the application plots for one simulation run.


```python
from datetime import datetime
from pathlib import Path
import re
import pandas as pd


def save_results(metricsTable, plotFigure, experimentName, resultsFolder):
    """Save the metrics CSV and the complete current GUI snapshot as a PNG."""

    if not isinstance(metricsTable, pd.DataFrame):
        raise TypeError("metricsTable must be a pandas DataFrame.")

    results_folder = Path(resultsFolder)
    results_folder.mkdir(parents=True, exist_ok=True)

    experiment_name = str(experimentName or "Experiment").strip() or "Experiment"
    experiment_name = re.sub(r"\s+", "_", experiment_name)
    experiment_name = re.sub(r'[<>:"/\\|?*]', "_", experiment_name)
    experiment_name = re.sub(r"_+", "_", experiment_name).strip("_") or "Experiment"

    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    base_name = f"{experiment_name}_{timestamp}"
    csv_file = results_folder / f"{base_name}_metrics.csv"
    png_file = results_folder / f"{base_name}.png"

    metricsTable.to_csv(csv_file, index=False)

    if plotFigure is None or not hasattr(plotFigure, "savefig"):
        raise ValueError("A valid Matplotlib figure is required to save the PNG.")

    plotFigure.savefig(png_file, dpi=200, bbox_inches="tight")


    return {
        "baseName": base_name,
        "csv": str(csv_file),
        "png": str(png_file),
    }
```
