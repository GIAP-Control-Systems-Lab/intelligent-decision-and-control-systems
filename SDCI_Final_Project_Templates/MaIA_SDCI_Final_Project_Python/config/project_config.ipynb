Defines the global project configuration used by the model and GUI.


```python
def project_config():
    """Return the shared project configuration."""

    # PROJECT
    CFG = {}
    CFG["projectName"] = "MaIA_SDCI"
    CFG["modelName"] = "MaIA_SDCI_Temperature_Control"

    # SIMULATION
    CFG["simulation"] = {}
    CFG["simulation"]["stopTime"] = 1800.0      # 30 min
    CFG["simulation"]["pacing"] = 100.0         # Simulation s / wall-clock s
    CFG["simulation"]["pacingEnabled"] = True

    # TEMPERATURE REFERENCES
    CFG["reference"] = {}
    CFG["reference"]["fourRooms"] = [22.0, 21.0, 20.0, 19.0]   # [degC]
    CFG["reference"]["singleRoom"] = 22.0                       # Room 5 / ESC [degC]

    # SIGNAL LOGGING
    CFG["logging"] = {}
    CFG["logging"]["enabled"] = True
    CFG["logging"]["signals"] = ["T_ref", "T_meas", "Q_heat"]

    # DISPLAY
    print("Project configuration loaded.")
    print(f"  Stop time:   {CFG['simulation']['stopTime']:.1f} s")
    refs = CFG["reference"]["fourRooms"]
    print(f"  References:  [{refs[0]:.1f} {refs[1]:.1f} {refs[2]:.1f} {refs[3]:.1f}] degC")
    print(f"  ESC ref.:    {CFG['reference']['singleRoom']:.1f} degC")

    return CFG
```
