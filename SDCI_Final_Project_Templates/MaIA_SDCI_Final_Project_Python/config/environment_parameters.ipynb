Defines the ambient temperature and external disturbance parameters.


```python
import math


def environment_parameters():
    """Return the environment and disturbance parameters."""

    # AMBIENT TEMPERATURE
    ENV = {}
    ENV["T_a"] = 8.0   # [degC]

    # TEMPERATURE DISTURBANCE
    ENV["disturbance"] = {}
    ENV["disturbance"]["enabled"] = False
    ENV["disturbance"]["amplitude"] = 13.0   # [degC]
    ENV["disturbance"]["startTime"] = 5.0    # [min]
    ENV["disturbance"]["endTime"] = 8.0      # [min]
    ENV["disturbance"]["roomMask"] = [1, 1, 1, 1]

    # VALIDATION
    if not isinstance(ENV["T_a"], (int, float)) or not math.isfinite(ENV["T_a"]):
        raise ValueError("Ambient temperature must be a finite scalar value.")

    disturbance = ENV["disturbance"]
    disturbance["enabled"] = bool(disturbance["enabled"])

    if not math.isfinite(float(disturbance["amplitude"])):
        raise ValueError("Disturbance amplitude must be a finite scalar value.")
    if not math.isfinite(float(disturbance["startTime"])) or disturbance["startTime"] < 0:
        raise ValueError("Disturbance start time must be a nonnegative finite scalar in minutes.")
    if not math.isfinite(float(disturbance["endTime"])) or disturbance["endTime"] < 0:
        raise ValueError("Disturbance end time must be a nonnegative finite scalar in minutes.")
    if disturbance["endTime"] <= disturbance["startTime"]:
        raise ValueError("Disturbance end time must be greater than start time.")
    if len(disturbance["roomMask"]) != 4:
        raise ValueError("Disturbance room mask must contain four elements.")
    if any(value not in (0, 1, False, True) for value in disturbance["roomMask"]):
        raise ValueError("Disturbance room mask values must be either 0 or 1.")

    disturbance["roomMask"] = [int(value) for value in disturbance["roomMask"]]

    # DISPLAY
    print("Environment parameters loaded.")
    print(f"  Ambient temperature:     {ENV['T_a']:.1f} degC")
    print(f"  Disturbance enabled:     {int(disturbance['enabled'])}")
    print(f"  Disturbance amplitude:   {disturbance['amplitude']:.1f} degC")
    print(f"  Disturbance start time:  {disturbance['startTime']:.1f} min")
    print(f"  Disturbance end time:    {disturbance['endTime']:.1f} min")
    print("  Disturbance room mask:   [{}]".format(" ".join(str(v) for v in disturbance["roomMask"])))

    return ENV
```
