Defines the sensor dynamics and measurement parameters.


```python
def sensor_parameters():
    """Return the sensor dynamics and measurement parameters."""

    SENSOR = {}

    # DYNAMICS
    SENSOR["tau"] = 10.0          # [s]
    SENSOR["sampleTime"] = 1.0    # [s]

    # INITIAL CONDITION
    SENSOR["initialTemperature"] = [8.0, 8.0, 8.0, 8.0]   # [degC]

    # MEASUREMENT IMPERFECTIONS
    SENSOR["bias"] = [0.0, 0.0, 0.0, 0.0]                 # [degC]
    SENSOR["noiseStd"] = [0.05, 0.05, 0.05, 0.05]         # [degC]
    SENSOR["seed"] = [101, 202, 303, 404]

    # DISPLAY
    print("Sensor parameters loaded.")
    print(f"  Time constant: {SENSOR['tau']:.2f} s")
    print(f"  Sample time:   {SENSOR['sampleTime']:.2f} s")
    print(f"  Noise std:     {SENSOR['noiseStd'][0]:.3f} degC")

    return SENSOR
```
