Defines the parameters for the equation-based actuator model.


```python
def actuator_parameters():
    """Return the heater and modulation parameters."""

    ACTUATOR = {}

    # ELECTRICAL SUPPLY
    ACTUATOR["V_rms"] = 240.0   # [V RMS]
    ACTUATOR["f_grid"] = 60.0   # [Hz]

    # RESISTIVE HEATER
    ACTUATOR["P_rated"] = 2500.0   # [W]
    ACTUATOR["R_heater"] = ACTUATOR["V_rms"] ** 2 / ACTUATOR["P_rated"]

    # MODULATION
    ACTUATOR["controlWindow"] = 1.0   # [s]
    ACTUATOR["cyclesPerWindow"] = round(ACTUATOR["f_grid"] * ACTUATOR["controlWindow"])

    # THERMAL CONVERSION
    ACTUATOR["efficiency"] = 1.0

    return ACTUATOR
```
