Defines the analytical thermal model used by the Python plant.


```python
import numpy as np


def plant_parameters():
    """Return the analytical four-room and single-room thermal models."""

    PLANT = {}

    # THERMAL PARAMETERS
    PLANT["C"] = 7.476e4       # [J/degC]
    PLANT["alpha"] = 123.6     # [W/degC]
    PLANT["Ta"] = 8.0          # [degC]

    # INITIAL CONDITIONS
    PLANT["initialTemperature"] = np.full(4, PLANT["Ta"], dtype=float)
    PLANT["initialTemperatureSingleRoom"] = float(PLANT["Ta"])

    # FOUR-ROOM MODEL
    Cth = PLANT["C"]
    alpha = PLANT["alpha"]

    A4 = np.array([
        [-2 * alpha / Cth,  alpha / Cth,                 0.0,                 0.0],
        [ alpha / Cth,     -3 * alpha / Cth,  alpha / Cth,                 0.0],
        [0.0,                alpha / Cth,     -3 * alpha / Cth,  alpha / Cth],
        [0.0,                       0.0,        alpha / Cth,     -2 * alpha / Cth],
    ], dtype=float)

    B4 = np.array([
        [1 / Cth, 0.0,     0.0,     0.0,     alpha / Cth],
        [0.0,     1 / Cth, 0.0,     0.0,     alpha / Cth],
        [0.0,     0.0,     1 / Cth, 0.0,     alpha / Cth],
        [0.0,     0.0,     0.0,     1 / Cth, alpha / Cth],
    ], dtype=float)

    PLANT["fourRooms"] = {
        "A": A4,
        "B": B4,
        "C": np.eye(4),
        "D": np.zeros((4, 5)),
        "initialTemperature": PLANT["initialTemperature"].copy(),
        "numberOfRooms": 4,
    }

    # SINGLE-ROOM MODEL
    PLANT["singleRoom"] = {
        "A": -alpha / Cth,
        "B": np.array([1 / Cth, alpha / Cth], dtype=float),
        "C": 1.0,
        "D": np.zeros((1, 2)),
        "initialTemperature": PLANT["initialTemperatureSingleRoom"],
        "numberOfRooms": 1,
    }

    return PLANT
```
