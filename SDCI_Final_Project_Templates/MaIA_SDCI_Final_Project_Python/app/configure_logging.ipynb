Defines the signals used by the GUI and post-processing utilities.


```python
def configure_logging(CFG):
    """Return the stable project signal names used by the GUI."""

    if not CFG.get("logging", {}).get("enabled", True):
        print("GUI signal logging disabled.")
        return []

    available_signals = {"T_ref", "T_meas", "Q_heat"}
    required_signals = list(CFG.get("logging", {}).get("signals", []))

    unknown = [signal for signal in required_signals if signal not in available_signals]
    if unknown:
        raise ValueError(f"Unknown logging signal(s): {', '.join(unknown)}")

    print("GUI signal logging configured.")
    for signal in required_signals:
        print(f"  {signal}")
    return required_signals
```
